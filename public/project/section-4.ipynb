# 1. Answering the Research Question

This project examined where Singapore's nightlife is concentrated and to what extent these areas experience gaps in public transport accessibility during late-night hours.

Combining OpenStreetMap (OSM) venue records, Land Transport Authority (LTA) Datamall transit networks, and October 2026 Origin-Destination (O-D) passenger flows reveals that Singapore's late-night mobility challenge is driven by **both spatial and temporal mismatches**:

1. **A concentrated downtown core paired with dispersed neighborhood hubs:** Across 620 mapped venues (72% pubs and bars, 18% nightclubs, dance clubs, and karaoke lounges, and 10% arcades and gaming centres), DBSCAN clustering identifies 7 distinct nightlife hotspots containing 61% of all establishments. The largest clusters sit in the Civic District and Bencoolen (165 venues) and along the Clarke Quay and Tanjong Pagar corridor (114 venues), alongside secondary hubs such as Geylang (52 venues). However, nearly 40% of venues (242 establishments) operate outside these clusters as dispersed neighborhood pubs and heartland karaoke lounges.
2. **Localized rail deserts and a post-midnight service cliff:** While average walking distances across all venues appear reasonable (491 m to an MRT station and 167 m to a bus stop), network analysis uncovers clear spatial gaps. Shophouse corridors like Joo Chiat Road and Balestier Road sit midway between major MRT lines, forcing patrons to walk over 800 m to reach a station. Temporally, venue openings drop sharply after 9:00 PM and plunge between 12:00 AM and 5:00 AM once MRT and regular bus services end, leaving patrons without affordable public transit options since the discontinuation of NightRider and NiteOwl services in 2022.

# 3. Policy Recommendations

Because reinstating an island-wide night bus network proved financially unsustainable in 2022, the spatial and O-D evidence points toward four targeted, lower-cost interventions across transit operations, street design, and urban zoning:

### 1. Reallocate Night Transit Along High-Volume O-D Corridors
* **Evidence:** Rather than uniform demand across the island, O-D flows show heavy uni-directional movement from NE5 Clarke Quay (122,873 trips) and Cluster 1 stations toward eastern Downtown Line hubs such as Tampines East (DT33) and Kaki Bukit (DT28).
* **Action:** Instead of reviving sprawling island-wide night routes, LTA can pilot a weekend and public-holiday late-night trunk bus corridor (12:00 AM to 4:00 AM) linking Clarke Quay and the Civic District directly to major eastern interchanges (Bedok, Tampines, and Pasir Ris). Focusing capacity on proven high-demand O-D corridors maximizes ridership efficiency while directly supporting Clarke Quay's 4:00 AM operating window.

### 2. Bridge Spatial Rail Deserts with First/Last-Mile Feeder Loops
* **Evidence:** OSMnx walking network analysis shows that venues along Joo Chiat Road and Balestier Road lie more than 800 m from the nearest MRT station, even though bus stops are located within 200 m of almost every establishment.
* **Action:** Partner with public transport operators to run short-loop feeder buses connecting rail-desert nightlife corridors (such as Joo Chiat to Paya Lebar/Eunos and Balestier to Novena/Toa Payoh) up to the final train departures around midnight, reducing missed-train friction without requiring new rail infrastructure.

### 3. Designate Time-Restricted Ride-Hail Pick-Up/Drop-Off (PUDO) Zones
* **Evidence:** Nearly 40% of nightlife venues sit outside major transit clusters, and private-hire vehicles remain essential for dispersed heartland patrons and post-midnight travelers. In dense shophouse districts like Tanjong Pagar, Clarke Quay, and Geylang, uncoordinated late-night ride-hail pickups create curbside bottlenecks and pedestrian safety hazards.
* **Action:** LTA could convert underutilized curbside parking bays in high-density clusters into dedicated, time-restricted pickup-dropoff zones between 11:00 PM and 4:00 AM, supported by clear pedestrian wayfinding from nightlife streets to main arterial roads.

### 4. Synchronize Nightlife Zoning and Licensing with Mobility Planning
* **Evidence:** Policy initiatives that extend nightlife vibrancy (e.g. formalizing 4:00 AM liquor licensing in Clarke Quay) are currently decoupled from public transport operating schedules, shifting external costs onto patrons via surge-priced private hire rides.
* **Action:** Establish a coordinated planning framework between URA (precinct planning), the Singapore Police Force (liquor licensing), and LTA so that designated late-night entertainment zones incorporate mobility impact assessments, shuttle co-funding models (with Business Improvement Districts), or targeted late-night transit provisions.

# 4. Methodological Limitations and Future Work

To ensure transparency and guide future spatial transit research, four technical limitations should be noted:

* **Sparse OSM Opening Hours Coverage:** Only 76 of the 620 venues  contained `opening_hours` metadata in OSM. Integrating commercial directory APIs would provide a more complete view of venue operating hours.
* **Cluster Boundary Edge Effects:** Strict spatial joins could exclude adjacent stations to clusters that add connectivity but are not captured by the cluster. Adding a 200m to 300m buffer around DBSCAN cluster convex hulls before intersecting with MRT nodes would better capture MRT stations serving these clusters.
* **Network Buffer Truncation and Indoor Linkways:** Capping the OSMnx pedestrian network query at a 1,000 m buffer left 11 venues with unreached MRT distances and 5 with unreached bus distances, while unmapped subterranean linkways inflated walking distances around integrated hubs like Marina Bay Sands. Expanding the network buffer to 3,000 m and incorporating LTA sheltered and underground pedestrian networks would refine walkability metrics.
* **Unimodal Rail Proxy and LRT Aggregation:** Because LTA separates MRT and bus O-D datasets, this study used MRT flows as a proxy for longer-distance commutes home, omitting bus-only (16.5%) and multimodal MRT-bus (24.8%) journeys. Additionally, treating LRT stops independently understates demand at parent MRT interchanges such as NE15 Buangkok and Sengkang/Punggol. Future work could aggregate LRT trips into their parent MRT hubs and merge bus O-D matrices for a fuller multimodal picture.
